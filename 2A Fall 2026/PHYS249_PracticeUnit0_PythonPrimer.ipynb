#### <font color=#003599>3.4 For Loops</font>

To make a set of code run repeatedly for a given set of values, **for** loops come in handy!

    r = [1,3,5]       # Created a list of values
    for n in r:       # For every value of n in the list r... repeat the following
        print(n)      # Print the value n for each value in the list
        print(2*n)    # Print 2*n for each value of n
    print("Finished") # Print "finished" when done loop
    
How can we run `for` loops?

- use a list `r` to run values through
- use `range(start_include, end_exclude, step_size)` like `range(2,6,1)` to generate a sequence of **integer** values from 2 to 6 (exclusive) with step sizes of 1
- use `arange(2,0,6.0,0.5)` to generate sequence of **float** values with step size 0.5
- use `linspace(2.0,5.0,100)` to generate a sequence of 100 values within the stated range

````python
# A for loop allows us to repeat an operation over a sequence of values
for i in range(5):
    print(i)
````

````python
# Useful commands for generating sequences of numbers include:
a = range(0,10)              # We're setting integers from 0 up to (but not including) 10
b = np.arange(0, 10, 1)      # We're getting values from first to last value (exclusive) in step-sizes of 1
c = np.linspace(0, 10, 100)  # We're creating 100 values from first to last value (inclusive)
print(a)
print(b)
print(c)
````

#### <font color=#118e07>EXAMPLES</font>
**Example 1**: Write a function `primes()` that takes one integer parameter and returns the number of primes less than that integer.

````python
def primes(n):
    '''Returns the number of primes less than the given integer'''
    count = 0
    for i in range(2, n):
        is_prime = True
        for j in range(2, int(i**0.5) + 1):
            if i % j == 0:
                is_prime = False
                break
        if is_prime:
            count += 1
    return count

print (primes(12))
````

### <font color=#003599>B1: Projectile Motion</font>

A ball is launched vertically upward with initial velocity

$$
v_0 = 20~\mathrm{m/s}.
$$

Ignoring air resistance, its height is

$$
y(t) = v_0t-\frac{1}{2}gt^2,
$$

where $g=9.81~\mathrm{m/s^2}$.

Write Python code that:

1. defines the physical constants;
2. creates 100 time values from $t=0$ to $t=4.1$ s;
3. calculates the corresponding heights;
4. plots height versus time with appropriate labels;
5. determines the approximate maximum height; and
6. identifies approximately when the maximum occurs.

> **Before running the code:** What do you expect the graph to look like? After running it, check whether your result makes physical sense.

````python
import matplotlib.pyplot as plt
#1. defines the physical constants;
#2. creates 100 time values from $t=0$ to $t=4.1$ s;
#3. calculates the corresponding heights;
#4. plots height versus time with appropriate labels;
#5. determines the approximate maximum height; and
#6. identifies approximately when the maximum occurs.
def projectile_motion(vo: float, t: float):
    g = 9.81
    vo = 20
    x= []
    y= []
    mh = 0 #max height during trajectory
    i = 0
    while i <= t:
        h = vo * i - 0.5 * g * i**2
        x.append(i)
        y.append(h)
        if h > mh:
            mh = h
            mt = i
        elif h < 0:
            x.pop()
            y.pop()
            print("Projectile has hit the ground at t = ", i, "s")
            break
        i += 0.025
    plt.plot(x, y)
    plt.xlabel("Time (s)")
    plt.ylabel("Height (m)")
    plt.title("Projectile Motion")
    plt.show()
    return mh, mt
print (projectile_motion(20, 5))
````

### <font color="#003599">B2. Experimental Motion Data</font>

An experiment records the position of an object at several times:

```python
time = np.array([0, 1, 2, 3, 4, 5])
position = np.array([0.2, 1.8, 4.1, 5.9, 8.2, 10.1])
```

Write Python code that:

1. plots the experimental data using `plt.scatter()`;
2. calculates the average velocity over each time interval using

$$
v = \frac{\Delta x}{\Delta t};
$$

3. prints the calculated velocities;
4. calculates their mean; and
5. describes in one or two sentences what kind of motion the data appear to represent.

> **Think first:** If the object is moving at approximately constant velocity, what should the position-versus-time graph look like?

````python
# Try this yourself!
#1. plots the experimental data using `plt.scatter()`;
#2. calculates the average velocity over each time interval using dx/dt
#3. prints the calculated velocities;
#4. calculates their mean; and
#5. describes in one or two sentences what kind of motion the data appear to represent.
time = np.array([0, 1, 2, 3, 4, 5])
position = np.array([0.2, 1.8, 4.1, 5.9, 8.2, 10.1])
plt.scatter(time, position)
plt.xlabel("Time (s)")
plt.ylabel("Position (m)")
plt.title("Experimental Data")
plt.show()
count = 0
velocity = []
for i in range(len(time)-1):
    dt = time[i+1] - time[i]
    dx = position[i+1] - position[i]
    velocity.append (dx/dt)
    print("Average velocity between t =", time[i], "and t =", time[i+1], "is", velocity, "m/s")
    count += 1

print("The mean velocity is", np.mean(velocity), "m/s")
````

**Interpretation:** The position increases ~linearly with time, and the calculated velocities are reasonably similar. The data therefore represents constant-velocity motion, with minor variation in measured values.

### <font color="#003599">C2. Debugging</font>

Errors are a normal part of programming. When something goes wrong, **read the error message!** Python will often tell you where it encountered the problem.

Some common things to check are:

- missing colons after `if`, `for`, `while`, or `def`;
- incorrect indentation;
- using `=` instead of `==`;
- misspelled variable or function names;
- missing parentheses or quotation marks; and
- using a variable before it has been defined.

Consider the following broken code:

```python
x = 5

if x = 5
print("x is five")

for i in range(5)
    print(i)
```

Can you identify the errors before looking at the error messages (good skill to have, especially for tests!).

````python
# Run this code and list out what errors there are - and fix them!
# With AI nowadays, there are these "Fix Code" buttons... refrain from using these! Use your brain instead!
x = 5

if x == 5:
    print("x is five")

for i in range(5):
    print(i)
````

````python
# There are several problems:
# 1. `=` assigns a value, while `==` checks whether two values are equal.
# 2. The `if` statement requires a colon (`:`).
# 3. The `print()` statement inside the `if` block must be indented.
# 4. The `for` statement also requires a colon (`:`).

# Now let's fix it
x = 5

if x == 5:
    print("x is five")

for i in range(5):
    print(i)
````

#### <font color="#00843D">Using `print()` to Debug</font>

Sometimes your program runs without an error but gives the **wrong answer**. In that case, print intermediate values to see what your program is actually doing.

> 💡 **PHYS 249 Tip:** If your final answer is wrong, don't just stare at the last line! Check the values produced at each important step of the calculation.

````python
# Something like this is useful 
x = 5
y = 3

result = x + y

print("x =", x)
print("y =", y)
print("result =", result)
````

### <font color="#003599">C3. Reading Files</font>

Scientific data are often stored in external files. NumPy provides convenient tools for loading numerical data.

For a simple text file:

```python
data = np.loadtxt("data.txt")
```

For a comma-separated file:

```python
data = np.loadtxt("data.csv", delimiter=",")
```

Here, `np.loadtxt()` loads numerical data into a NumPy array, while `delimiter` tells Python how the values in the file are separated.
