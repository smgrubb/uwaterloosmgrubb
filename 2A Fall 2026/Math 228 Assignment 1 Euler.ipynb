#Euler's method solver

This is a python program I (Shintaro) wrote so that all of you can use Euler's method to solve 1st order initial value problems. I will work with placeholder functions, but hopefully you can see how to modify the code to solve *any* first order IVP.

General IVP: $$y'(x) = f(x,y(x)),$$
$y(x_0) = y_0.$ There is no catch-all analytic method to solve this. However, we can approximate solutions with Euler's method.

Our example: $$y′(x)=\sin(y)^{\cos(x)}, y(0)=1.$$ Problem: Estimate $y(5)$.

(The true value is roughly 7.1187.)

```python
import math
```

# Initializing everything
Initial values and step size are implemented here, as well as the target $x$ value.

```python
x_0 = 0 #Declare initial x value x_0
x_N = 5 #Declare target x value x_N
y_0 = 1 #Declare initial value y(x_0)
h = 1 #Declare step size.
```

#Implementation of Euler's method

Everything is now initialized. It remains to "run the loop".

```python
x_i = x_0 #Initial point for iteration
y_i = y_0 #Initial value for iteration

steps = 0 #Initially we have done no steps.
totalsteps = round((x_N-x_0)/h) #The number of steps needed to reach x_N

while steps < totalsteps: #Loop enough times to reach x_N.
    gradient_i = abs(math.sin(y_i))**math.cos(x_i) #The ODE. ######UPDATE THIS AS NEEDED.
    x_i1 = x_i + h
    y_i1 = y_i + h*gradient_i
    x_i = x_i1
    y_i = y_i1
    steps += 1

print(x_i, y_i)
```

$\frac{dy(t)}{dt}=2y(t)t$, with initial value $y(0)=0.1$

```python
def EulerMethod(x_0, x_N, y_0, h):
    x_i = x_0
    y_i = y_0
    steps = 0
    totalsteps = round((x_N-x_0)/h)
    
    while steps < totalsteps:
        gradient_i = 2*y_i*x_i #The ODE. ######UPDATE THIS AS NEEDED.
        x_i1 = x_i + h
        y_i1 = y_i + h*gradient_i
        x_i = x_i1
        y_i = y_i1
        steps += 1

    return x_i, y_i
```

```python
def ImprovedEulerMethod(x_0, x_N, y_0, h):
    x_i = x_0
    y_i = y_0
    steps = 0
    totalsteps = round((x_N-x_0)/h)
    
    while steps < totalsteps:
        gradient_i = 2*y_i*x_i #The ODE. ######UPDATE THIS AS NEEDED.
        x_i1 = x_i + h
        y_predictor = y_i + h*gradient_i #Predictor step
        gradient_predictor = 2*y_predictor*x_i1 #Evaluate gradient at predicted point
        g = (gradient_i + gradient_predictor)/2 #Average the gradients
        y_i1 = y_i + (h*g) #Corrector step
        x_i = x_i1
        y_i = y_i1
        steps += 1

    return x_i, y_i
```

```python
t = 2
y = 0.1 * math.e**(t**2) # I wanted to have the errors to help me discuss any observations about the errors produced by the different methods.
print ('Exact value of y(2):', y)
print ('Estimate of y(2) with step size 0.1:', EulerMethod(0, 2, 0.1, 0.1)[1], 'with error of', abs(EulerMethod(0, 2, 0.1, 0.1)[1] - y))
print ('Estimate of y(2) with step size 0.05:', EulerMethod(0, 2, 0.1, 0.05)[1], 'with error of', abs(EulerMethod(0, 2, 0.1, 0.05)[1] - y))
print ('Improved Euler Estimate of y(2) with step size 0.1:', ImprovedEulerMethod(0, 2, 0.1, 0.1)[1], 'with error of', abs(ImprovedEulerMethod(0, 2, 0.1, 0.1)[1] - y))
```
