Implement Euler's method with a step-size of ℎ=0.1 to approximate 𝑦(10) for the following two IVPs. You should submit evidence of your implementation (e.g. a screenshot of your code), as well as explicitly stating your approximation of 𝑦(10) for each IVP.
𝑦″(𝑥)+𝑥𝑦(𝑥)=0,𝑦(0)=1,𝑦′(0)=0
𝑦″(𝑥)−𝑥𝑦(𝑥)=0,𝑦(0)=1,𝑦′(0)=0
For which IVP do you expect the approximation to be closer to the true value of 𝑦(10)? Explain your reasoning.

```python
# Euler's method for a second-order ODE y'' = func(x, y, y_prime).
def Euler(func, h, x_0, y_0, x_t, y_prime_0):
    """
    Approximate y and y' from x_0 to x_t using Euler's method.

    func takes the current x, y, and y' and returns y''.
    Returns the final x, y, and y' values.
    """
    x_i = x_0 #Set the variables
    y_i = y_0
    y_prime_i = y_prime_0
    steps = 0
    totalsteps = round((x_t - x_0) / h)

    while steps < totalsteps:
        y_double_prime_i = func(x_i, y_i, y_prime_i) #Calculates the 2nd derivative
        x_i1 = x_i + h
        y_i1 = y_i + h * y_prime_i
        y_prime_i1 = y_prime_i + h * y_double_prime_i #Calculates the first derivative
        x_i = x_i1 #Updates the variables
        y_i = y_i1
        y_prime_i = y_prime_i1
        steps += 1

    return x_i, y_i, y_prime_i


# y'' + x*y = 0, so y'' = -x*y.
y_at_10_first = Euler(lambda x, y, y_prime: -x * y, 0.1, 0, 1, 10, 0) #Lambda to create the function
print("Approximation of y(10) for y'' + x*y = 0:", y_at_10_first[1])

# y'' - x*y = 0, so y'' = x*y.
y_at_10_second = Euler(lambda x, y, y_prime: x * y, 0.1, 0, 1, 10, 0)
print("Approximation of y(10) for y'' - x*y = 0:", y_at_10_second[1])
```
